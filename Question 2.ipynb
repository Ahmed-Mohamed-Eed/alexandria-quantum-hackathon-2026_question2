```
import numpy as np

from qiskit.quantum_info import Statevector, Pauli
from qiskit.visualization import plot_bloch_vector

alpha = np.sqrt(7/12)

beta = np.sqrt(5/12) * np.exp(1j * np.pi/4)

psi = Statevector([alpha, beta])

x = np.real(psi.expectation_value(Pauli("X")))
y = np.real(psi.expectation_value(Pauli("Y")))
z = np.real(psi.expectation_value(Pauli("Z")))

print("x =", x)
print("y =", y)
print("z =", z)

plot_bloch_vector([x, y, z])
```
